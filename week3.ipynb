```python
# Archetype 1: The Sieve (Filtering)Problem: You are given a list of test scores: scores = [45, 82, 30, 91, 58, 76]. Extract only the passing scores ($60$ or higher) into a new list.

# 1.the sieve or filtering
scores = [45, 82, 30, 91, 58, 76]

filtered_score = []

for score in scores:
    if score > 60:
        filtered_score.append(score)

print(filtered_score)

# 2.Archetype 2: The Accumulator (Tallying / Totals) -piggy bank

# Problem: You are given a list of weekly grocery receipts: receipts = [12, 45, 8, 23]. Calculate the grand total spent.

receipts = [12, 45, 8, 23]
total = 0

for reciept in receipts:
    total = total + reciept

print(total)

# Archetype 3: The King of the Hill (Min / Max)
# Problem: You are given daily temperatures recorded over a week: temps = [68, 74, 61, 85, 79]. Find the hottest day without using max().
temperatures_data = [68, 74, 61, 85, 79]
king = temperatures_data[0]

for temperature in temperatures_data:
    if temperature > king:
        king = temperature

print(king)


# Archetype 4: The Scout (Searching / Early Exit)
# Problem: You have a list of server status codes: codes = [200, 200, 500, 200, 404]. Determine if there is at least one critical server crash (500). Answer True or False.

codes = [200, 200, 500, 200, 404]
found = False

for code in codes:
    if code == 500:
        found = True
        break

```

## Topic 5: Loops

1. Calculate total revenue from a list of order dictionaries.
2. Count paid orders and collect the IDs of unpaid orders.
3. Validate records in a loop and collect IDs whose amount is missing or non-positive.

```python
# Answers 1-3
# Write your answers here
orders = [
    {"order_id": 101, "amount": 150.00, "status": "paid"},
    {"order_id": 102, "amount": 0.00, "status": "unpaid"},
    {"order_id": 103, "amount": 75.50, "status": "unpaid"},
    {"order_id": 104, "amount": None, "status": "paid"},
    {"order_id": 105, "amount": -20.00, "status": "pending"},
    {"order_id": 106, "status": "unpaid"},
    {"order_id": 107, "amount": 200.00, "status": "paid"}
]

total_revenue = 0.0
paid_count = 0
unpaid_order_ids = []
invalid_order_ids = []

for order in orders:
    order_id = order.get("order_id")
    amount = order.get("amount")
    status = order.get("status")

    if amount is None or amount <= 0:
        invalid_order_ids.append(order_id)
    else:
        total_revenue += amount

    if status == "paid":
        paid_count += 1
    else:
        unpaid_order_ids.append(order_id)

print(total_revenue)
print(paid_count)
print(unpaid_order_ids)
print(invalid_order_ids)
```

## Topic 6: Functions

1. Write `net_price(price, discount_rate=0)` with validation for a rate from 0 to 1.
2. Write `active_employee_names(records)` that returns names for active employees.
3. Write `summarize_values(values)` returning count, total, and average.

```python
# Answers 1-3
# Write your answers here

def net_price(price, discount_rate=0):
    if not (0 <= discount_rate <= 1):
        raise ValueError("discount_rate must be between 0 and 1")
    return price * (1 - discount_rate)


def active_employee_names(records):
    names = []
    for record in records:
        if record.get("status") == "active":
            names.append(record.get("name"))
    return names


def summarize_values(values):
    count = len(values)
    if count == 0:
        return 0, 0, 0
    total = sum(values)
    average = total / count
    return count, total, average
```

### Question 3
Count how many values in `[5, 12, 8, 20]` are greater than 10.

```python
# Answer 3
# Write your answer here


ls = [5, 12, 8, 20]
count = 0

for i in ls:
    if i > 10:
        count = count + 1

print(count)
```
